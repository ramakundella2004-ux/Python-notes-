```python
# upper part

for i in range(1,6,1):

    j=5
    while(j>i):
        print(" ",end="")
        j-=1

    k=1
    while(k<=i):
        if(i==1 or k==1 or k==i):
            print("*",end=" ")
        else:
            print(" ",end=" ")
        k+=1

    print()

# lower part
for i in range(1,5,1):

    j=1
    while(j<=i):
        print(" ",end="")
        j+=1

    k=4
    while(k>=i):
        if(i==4 or k==4 or k==i):
            print("*",end=" ")
        else:
            print(" ",end=" ")

        k-=1

        
    print()
            
```
