```python
# password strength checker 
pw=input("enter your password:")
uc=0
lc=0
sp=0
nd=0
if (len(pw)>=8):
    for i in pw:
        if i.isupper():
            uc+=1
        elif i.islower():
            lc+=1
        elif i.isdigit():
            nd+=1
        else:
            sp+=1
else:
    print("password length is more that 7 charecters")

if(uc>=1 and lc>=1 and nd>=1 and sp>=1):
    print("strong password")
else:
    print("weak password")
    
```

```python
n = int(input("Enter a number: "))

even_sum = 0
odd_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print("Sum of Even Numbers:", even_sum)
print("Sum of Odd Numbers:", odd_sum)
```

```python
numbers = [10, 45, 23, 89, 67]

largest = float('-inf')
second = float('-inf')

for num in numbers:
    if num > largest:
        second = largest
        largest = num

    elif num > second:
        second = num

if len(numbers) < 2:
    print("At least two numbers are required")
else:
    print("Second Largest Number:", second)
```

```python
n = abs(int(input("Enter a number: ")))

count = 0

for i in range(n, 0, -1):
    count += 1
    n //= 10

    if n == 0:
        break

if count == 0:
    count = 1

print("Number of digits:", count)
```

```python
text = input("Enter a string: ")

vowels = 0
consonants = 0

for ch in text.lower():
    if ch.isalpha():
        if ch in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
n = int(input("Enter a positive number: "))

factorial = 1

if n < 0:
    print("Factorial is not defined for negative numbers")
else:
    for i in range(1, n + 1):
        factorial *= i

    print("Factorial:", factorial)
```
