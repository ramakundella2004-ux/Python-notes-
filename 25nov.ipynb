```python
write a code for prime number and even and odd from the given range
```

```python
sv= int(input("enter a start number:"))
st=int(input("enter a stop number:"))
pl=[]
en=[]
on=[]
for i in range(sv,st+1,1):
    if i%2==0:
       en.append(i)
    else:
        on.append(i)
    
    if(i==1):
        continue
    for j in range(2,i,1):
        if i%j==0:
            break
    else:
        pl.append(i)

    
print(f"prime numbers:{pl}\neven number:{en}\nodd number:{on}")

    
```

```python
st="My Name is Sharmila, Iam 21 year old"
uc=[]
lc=[]
sp=[]
nl=[]
for i in st:
    if i ==" ":
        continue
    if i.isupper():
      uc.append(i)
    elif i.islower():
        lc.append(i)
    elif i.isdigit():
        nl.append(i)
    else:
        sp.append(i)

print(f"uppercase:{uc}\nlowercase:{lc}\nspecial charater:{sp}\nnumber:{nl}")
```
