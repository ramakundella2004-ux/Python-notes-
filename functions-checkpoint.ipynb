```python
def prime1():
    stv=int(input("enter stop value:"))
    spv=int(input("enter start value:"))
    for i in range(stv,spv+1,1):
        if (i==1):
            continue
        for j in range(2,i,1):
              if i%j==0:
                  break
        else:

            print(i)
```

```python
lambda function
a lambda function is a small, anonymous(nameless)
function in python

* it defined using keywords lambda insted of def
* it can take any number of arguments but must containe only one expression
* expression is automatically returned (no need to return)

syntax
lambda arguments:expression

why to use lambda?
* useful for short,simple operation
* saves code(one-liner)
* often used with map(),filter(),reduce() or as inline function

when not use lambda?
*
for complex logic(use def insted)
* when redability is more important than shortness.
```

```python
s=lambda a,b:a+b
```

```python
s(4,6)
```
